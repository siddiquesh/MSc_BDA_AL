```python
#Byesian Updating for medical Testing

#prior probability
prior_dieases = 0.01

#test characteristics
sensitivity = 0.95
specificity = 0.90

#probability of positive test among people without disease
false_positive_rate = 1 - specificity

#Bayes's theorem fter first positive test
prob_positive = ( sensitivity * prior_dieases ) + ( false_positive_rate * (1 - prior_dieases))

posterior_1 =( sensitivity * prior_dieases ) / prob_positive

print("probability of positive test: ",prob_positive)
print("posterior proability after first positive test:",posterior_1)
```

Q2)

```python
#use the posterior probability after first test as the new prior
new_prior = posterior_1

prob_positive_2 = ( sensitivity * new_prior ) + ( false_positive_rate * (1 - new_prior))

posterior_2 =( sensitivity * new_prior ) / prob_positive_2

print("probability of positive test: ",prob_positive_2)
print("posterior proability after second positive test:",posterior_2)
```
