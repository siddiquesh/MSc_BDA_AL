```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

Q1) Generate values of (X) from 20 to 120 and calculate the probability density function of

 # **X∼N(70,102).**

Plot the Normal distribution curve.
Observe the graph and state:

1. Is the distribution symmetric?
2. Around which value is the curve centred?
1. What happens to the spread if standard deviation is increased?





```python
# Q1
mu=70
sigma =10
x= np.linspace(20,120,500)
pdf = stats.norm.pdf(x,loc=mu,scale=sigma)
plt.plot(x,pdf)
plt.xlabel("X")
plt.ylabel("Normal Distribution")
plt.grid()
plt.show()
```

1. yes, it is symmetric
2. At 70 the curve is centered
3. that the curve become more spread

* Q2) Plot three Normal distributions having the same mean=70 but different List item standard deviations 5,10,15.
*  Use the same graph to compare their shapes.

Observe the graph and answer:

1. Which distribution is the most concentrated?
2. Which distribution has the greatest spread?
1. What happens to the height of the curve as standard deviation increases?


```python
# Q2
mu=70

x= np.linspace(20,120,500)

pdf = stats.norm.pdf(x,loc=mu,scale=5)
pdf2 = stats.norm.pdf(x,loc=mu,scale=10)
pdf3 = stats.norm.pdf(x,loc=mu,scale=15)

plt.figure(figsize=(9,5))

plt.plot(x,pdf,label="sigma=5")
plt.plot(x,pdf2,label="sigma=10")
plt.plot(x,pdf3,label="sigma=15")

plt.xlabel("Marks")
plt.ylabel("Probability Density")
plt.title("Effect of Standard Deviation on Normal Distribution")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

1. Most concentrated: The distribution with sigma=5.
2. Greatest spread: The distribution with sigma=15.
3. Height of the curve: It decreases, making the curve flatter.

Q3) Plot three Normal distributions having the same standard deviation but different means:
N(50,102),N(70,102),N(90,102).
1. What happens to the Normal distribution when the mean changes while the standard deviation remains constant?


```python
# Q3
sigma=10

x= np.linspace(20,120,500)

pdf = stats.norm.pdf(x,loc=50,scale=sigma)
pdf2 = stats.norm.pdf(x,loc=70,scale=sigma)
pdf3 = stats.norm.pdf(x,loc=90,scale=sigma)

plt.figure(figsize=(9,5))

plt.plot(x,pdf,label="sigma=5")
plt.plot(x,pdf2,label="sigma=10")
plt.plot(x,pdf3,label="sigma=15")

plt.xlabel("Marks")
plt.ylabel("Probability Density")
plt.title("Effect of Standard Deviation on Normal Distribution")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

1. The entire distribution shifts horizontally along the x-axis without changing its shape or spread.

Q4) Consider a situation where a customer either:
purchases a product, or does not purchase a product.
Suppose the probability of purchase is
p=0.3

1. If 20 customers are observed, then calculate and plot the probability mass function.

```python
#Q4

n=20
p = 0.3
x=np.arange(0,n+1)
pmf=stats.binom.pmf(x,n,p)
plt.stem(x,pmf)
plt.xlabel("Number of Purchases")
plt.ylabel("Probability")
plt.title("Binomial Distribution")
plt.grid()
plt.show()
```

Q5) A customer-support centre receives an average of 4 calls per minute.
Model the number of calls received in one minute using a Poisson distribution and plot its probability mass function.
1. What is the value of lambda?
2. Is the Poisson distribution discrete or continuous?
3. What is the expected number of calls per minute?

```python
# Q5
lam=4
x=np.arange(0,20)
pmf=stats.poisson.pmf(x,lam)
plt.stem(x,pmf)
plt.xlabel("Number of Cars")
plt.ylabel("Probability")
plt.title("Poisson Distribution")
plt.grid()
plt.show()
```

1. lamba= 4
2. distribution is discrete
3. the expected number of call is 4
