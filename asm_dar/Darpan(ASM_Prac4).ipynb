```python
import seaborn as sns
from scipy import stats

print("H0 = the mean fare paid by male and female passenger differs significantly")
print("H1 = the mean fare paid by male and female passenger differs in significantly")

df= sns.load_dataset("titanic")

male = df[df["sex"] == "male"]["fare"].dropna()
female = df[df["sex"] == "female"]["fare"].dropna()

t_stat, p_value = stats.ttest_ind(
    male,female,equal_var=False
)
print("t-static :", t_stat)
print("p_value :", p_value)
```

p_value < 0.05 , hence reject the H0 , the mean fare paid by male and female passenger differs in significantly

```python
from scipy.stats import chi2_contingency
import pandas as pd

print("H0:- the passenger class and the survival status are not associated")
print("H1:- the passenger class and the survival status are associated")

table = pd.crosstab(df["class"],df["survived"])
print(table)

chi2, p_value, dof, expected = chi2_contingency(table)

print("Chi square statistic :",chi2)
print("Degree of freedom :",dof)
print("p-value :",p_value)
```

p-value < 0.05 , hence reject the H0 ,the passenger class and the survival status are associated

```python
print("H0:- mean fares do not differs significantly across 3 passenger class")
print("H1:- mean fares differs significantly across 3 passenger class")
from scipy.stats import f_oneway
first = df[df["class"] == "First"]["fare"].dropna()
second = df[df["class"] == "Second"]["fare"].dropna()
third = df[df["class"] == "Third"]["fare"].dropna()

F, p_value = f_oneway(first,second,third)

print("F-statistic: ", F)
print("p_value: ",p_value)
```
