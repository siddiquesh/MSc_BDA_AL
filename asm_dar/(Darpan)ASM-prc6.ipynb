```python
from sklearn.datasets import load_diabetes
import pandas as pd
import statsmodels.api as sm

diabetes = load_diabetes(as_frame=True)
df = diabetes.frame
df.head()
```

Q1)

```python
#correlation
print("--- Correlation ---")
print(df[['bmi',"target"]].corr())

#simple linear regression
x=sm.add_constant(df['bmi'])
y=df['target']
model=sm.OLS(y,x).fit()
print("\n--- Regression Summary ---")
print(model.summary())
```

Slope interpretation: For every one-unit increase in BMI, predicted Y increases by 949.4353 units on average.

Q2)

```python
x = df[["bmi","bp","age"]]
x = sm.add_constant(x)
y = df["target"]
model = sm.OLS(y,x).fit()
print(model.summary())
```

### Regression Coefficients Summary

* **Significant Predictors (p < 0.05):**
  * **BMI** (`coef = 788.78`): Strongest positive effect. A 1-unit increase in normalized BMI yields a ~788.8 unit increase in the target.
  * **BP** (`coef = 394.13`): Positive effect. A 1-unit increase in normalized blood pressure yields a ~394.1 unit increase in the target.
  * **Intercept (const)** (`coef = 152.13`): Expected value of target when all predictors are zero.

* **Insignificant Predictors (p >= 0.05):**
  * **Age** (`coef = 25.99`, `p = 0.685`): No statistically significant relationship with the target in this model.

```python
from scipy.stats import spearmanr

study_time = [2,4,1,5,3]
exam_score = [50,65,45,80,60]

correlation, p_value = spearmanr(study_time, exam_score)

print("Spearman's correlation coefficient:", correlation)
print("p-value:", p_value)
```

Spearman correlation coefficient is Perfectly Positievly Correlated
